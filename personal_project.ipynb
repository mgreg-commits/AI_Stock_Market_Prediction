```python
import pandas as pd
import os
import ta  # lightweight technical analysis library

# Step 1: Merge all CSV files into one DataFrame
data_folder = 'top50_data'
all_data = []

for file in os.listdir(data_folder):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(data_folder, file))
        ticker = file.replace('.csv', '')
        df['Ticker'] = ticker  # Add ticker as a column
        all_data.append(df)

# Concatenate all into a single DataFrame
stock_data = pd.concat(all_data, ignore_index=True)

# Step 2: Convert 'date' to datetime and sort
stock_data['date'] = pd.to_datetime(stock_data['date'])
stock_data.sort_values(by=['Ticker', 'date'], inplace=True)

# Step 3: Clean column names for consistency
stock_data.rename(columns={
    '1. open': 'Open',
    '2. high': 'High',
    '3. low': 'Low',
    '4. close': 'Close',
    '5. volume': 'Volume'
}, inplace=True)

# Step 4: Apply technical indicators
# We'll apply indicators group-by-ticker
def apply_indicators(df):
    # SMA and EMA
    df['SMA_20'] = ta.trend.sma_indicator(df['Close'], window=20)
    df['EMA_20'] = ta.trend.ema_indicator(df['Close'], window=20)
    
    # RSI
    df['RSI_14'] = ta.momentum.rsi(df['Close'], window=14)

    # MACD
    df['MACD'] = ta.trend.macd(df['Close'])
    df['MACD_signal'] = ta.trend.macd_signal(df['Close'])

    # Bollinger Bands
    bb = ta.volatility.BollingerBands(df['Close'], window=20, window_dev=2)
    df['BB_upper'] = bb.bollinger_hband()
    df['BB_lower'] = bb.bollinger_lband()
    
    return df

# Group by each ticker and apply indicators
stock_data = stock_data.groupby('Ticker').apply(apply_indicators).reset_index(drop=True)

# Step 5: Save to processed CSV
stock_data.to_csv("processed_stock_data.csv", index=False)

# Preview
print(stock_data[['Ticker', 'date', 'Close', 'SMA_20', 'RSI_14']].head())
```

```python
import pandas as pd
from sklearn.linear_model import Lasso, Ridge
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split
import numpy as np

# Step 1: Load processed stock data
df = pd.read_csv("processed_stock_data.csv")

# Step 2: Create a next-day target column (Close shifted -1)
df['Close_tomorrow'] = df.groupby('Ticker')['Close'].shift(-1)

# Step 3: Drop rows with any missing values (from indicator calculations and shifting)
df.dropna(inplace=True)

# Step 4: Define feature columns (indicators only)
features = ['SMA_20', 'EMA_20', 'RSI_14', 'MACD', 'MACD_signal', 'BB_upper', 'BB_lower']

# Step 5: Define X and y
X = df[features]
y = df['Close_tomorrow']

# Step 6: Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=42)

# Step 7: Initialize models
models = {
    'Lasso': Lasso(alpha=0.1, max_iter=10000),
    'Ridge': Ridge(alpha=1.0),
    'XGBoost': GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=42)
}

# Step 8: Train and evaluate each model
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    
    mae = mean_absolute_error(y_test, preds)
    mape = np.mean(np.abs((y_test - preds) / y_test)) * 100
    rmse = mean_squared_error(y_test, preds) ** 0.5

    
    results[name] = {
        'MAE': mae,
        'MAPE': mape,
        'RMSE': rmse
    }

# Step 9: Display performance
print("📊 Model Performance:")
for name, metrics in results.items():
    print(f"\n{name} Regression:")
    for metric, value in metrics.items():
        print(f"  {metric}: {value:.4f}")
```

```python
import gym
from gym import spaces
import numpy as np

class MultiStockTradingEnv(gym.Env):
    def __init__(self, df, ticker):
        super(MultiStockTradingEnv, self).__init__()
        self.df = df[df['Ticker'] == ticker].reset_index(drop=True)
        self.ticker = ticker
        self.initial_balance = 10000
        self.reset()

        # Action: 0=Hold, 1=Buy, 2=Sell
        self.action_space = spaces.Discrete(3)

        # Observation: 7 indicators + Close + balance + position
        self.observation_space = spaces.Box(
            low=-np.inf, high=np.inf, shape=(10,), dtype=np.float32
        )

    def _get_obs(self):
        row = self.df.loc[self.current_step]
        obs = np.array([
            row['SMA_20'], row['EMA_20'], row['RSI_14'],
            row['MACD'], row['MACD_signal'],
            row['BB_upper'], row['BB_lower'],
            row['Close'], self.balance, self.position
        ], dtype=np.float32)
        return obs

    def reset(self):
        self.current_step = 0
        self.balance = self.initial_balance
        self.position = 0
        self.net_worth = self.initial_balance
        return self._get_obs()

    def step(self, action):
        current_price = self.df.loc[self.current_step, 'Close']

        # Buy
        if action == 1 and self.balance >= current_price:
            self.position += 1
            self.balance -= current_price

        # Sell
        elif action == 2 and self.position > 0:
            self.position -= 1
            self.balance += current_price

        # Advance time
        self.current_step += 1
        done = self.current_step >= len(self.df) - 1
        self.net_worth = self.balance + self.position * current_price
        reward = self.net_worth - self.initial_balance

        return self._get_obs(), reward, done, {}

    def render(self, mode='human'):
        print(f"[{self.ticker}] Step: {self.current_step} | "
              f"Balance: {self.balance:.2f} | Position: {self.position} | "
              f"Net Worth: {self.net_worth:.2f}")
```
